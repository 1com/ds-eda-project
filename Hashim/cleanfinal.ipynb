Data Cleaning Step 1: Convert column date to datetime format

```python
df_housing["date"] = pd.to_datetime(df_housing["date"], format="%Y-%m-%d")
```

Note on `bathrooms` decimals:
Decimal values in `bathrooms` etc are not data errors — they follow a standard US real estate convention where each fixture (toilet, sink, shower, tub) counts as 0.25. For example, 0.5 = half bath, 0.75 = three-quarter bath, and 1.75 = one full bathroom + one three-quarter bathroom. We will leave these values unchanged.

Data Cleaning Step 2: Column yr_renovated contains 3858 (17.8%) NaN values. which we are interpreting as no renovation as no date was mentioned for renovation, therefore replacing with 0 and converting to INT data type to get rid of the decimals. Additionally this column contains an extra 0 at the end of all years which needs to be removed by dividing by 10.

```python
df_housing.info()
```

```python
df_housing["yr_renovated"].isna().sum()
```

```python
df_housing["yr_renovated"].info()
```

```python
df_housing["yr_renovated"]
```

Data Clean Step 4: Fill NA for view. Same logic as waterfront because only 63 values from view are missing and the data shows around 90% of the houses have 0 for a view. For simplicity we also convert this to int

```python
df_housing["view"] = df_housing["view"].fillna(0)
df_housing["view"] = df_housing["view"].astype(int)
df_housing["view"].value_counts(normalize=True)
```

Data Clean Step 5: Replace missing values for the sqft_basement column. First we verify if there is a relationship between the different sqft values such as sqft_basement = sqft_living - sqft_above

```python
df_housing.isna().sum()
```

```python
known = df_housing.dropna(subset = ["sqft_basement"])
(known["sqft_living"] - known["sqft_above"] == known["sqft_basement"]).all()
```

```python
maskh = df_housing["sqft_basement"].isna()
df_housing.loc[maskh, "sqft_basement"] = (
    df_housing.loc[maskh, "sqft_living"] - df_housing.loc[maskh, "sqft_above"]
)
```
